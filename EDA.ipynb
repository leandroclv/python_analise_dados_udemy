###Introdução à análise de dados

A análise de dados é uma área que envolve a prática constante de transformar dados brutos em informações úteis, visando a melhoria de negócios e a melhoria de performance das empresas frente às transformações digitais que ocorrem atualmente.

Esta época é chamada de Era da Informação, onde os dados são considerados o novo petróleo, pois a partir deles é possível gerar informações que vão impcatar diretamente na competividade e no sucesso da empresa frente ao mercado. Por isso, atualmente, o processo organizacional como um todo é voltado e orientado a dados

Com a tranformação digital e o foco nos dados, as empresas passaram a lidar com o tratamento de grandes volumes de dados, gerados e coletados em alta velocidade e a parir de diversas fontes e tipos. Este fenômeno é conhecido como Big Data.

Para que os dados possam ser analisados, são utilizados diversos conceitos fundamentais de de computação e matemática, os mesmos que formam em conjunto os conceitos gerais da ciÇencia e mineração de dados.

Para entendermos melhor esses conceitos, temos que esclarecer algumas termologias básicas utilizadas comumente. São elas:

####Ordem de significado

1. Dados: consiste em fatos discretos que representam a base da informação, porém ainda não possuem significado qualquer;
2. Informação: consiste na organização e estruturação dos fatos, aplicando contexto e dando significado aos dados;
3. Conhecimento: consiste em informações processadas e transformadasem experiência por uma pessoa.

####Tipos de dados

- Estruturados: dados com esrígida e previamente definida, formados basicamente por tabelas de banco de dados;
- Não Estruturados: dados que não requerem estrutura bem definidas ou padronizadas, mais comuns no dia a dia das pessoas, formados por textos, fotos, vídeos, áudios, etc;
- Semi-estruturados: dados que possuem uma combinação de ambas as características, pois ao mesmo tempo que são de, alguma forma, organizados, não são tão rígidos como os dados estruturados formados por códigos em HTML, JSON entre outros.

###Etapas fundamentais

O processo da análise de dadospode ser dividido em três grandes etapas. São elas:

1. Levantamento: é a parte da análise onde os dados são coletados e é realizado o entendimento inicial da sua natureza e do negócio em si. Em outras palavras, visa se responder 'do que se tratam os dados'.

2. Exploração: é a parte da análise onde ocorre o aprofundamento e foca no significado e relação entre os dados. Em outras palavras, visa responder 'o que os dados dizem e/ou sugerem?

3. Mineração: é a parte da análise onde ocorre um processamento sistemático e metódico para a identificação de padrões e informações implícitas dentro dos dados. Em outras palavras, visa se responder 'Como extrair informações úteis dos dados?'

####Observações

- As etapas mencionadas acima também podem ser relacionadas com as terminologias de significado

- Essas etapas são apenas uma proposta lógica de aplicações comumente utilizadas durante o processo de análise de dados. Porém, podem existir outras variações.

- Existe também a etapa decomunicação dos resultados para o cliente ou demais áreas. Esta ocorre ao final da análise completa através de relatórioa e paineis interativos, onde são utilizadas diversas ferramentass e concetios analíticos de BI, ligados ao negócio.

####Tipos de análise

A análise de dados pode ser dividida em 4 tipos diferentes. São eles:

1. Descritiva: visa responder 'o que aconteceu?'
2. Diagnóstica: visa responder 'por que aconteceu?'
3. Preditiva: visa responder 'o que vai acontecer?'
4. Prescritiva: visa responder 'o que fazer quando acontecer?'

####Aprendizado de máquina

O aprendizado de máquina, ou Machine Learning, é uma área dentro da análise de dados que utiliza diversos algoritmos, técniicas de matemática e conceitos estatísticos para criar modelos de indução a partir dos dados coletados e analisados.

Para sua execução, são utilizados diversas etapas e processamentos intermediários. De modo geral, é possível utilizar o seguinte pipeline para o seu funcionamento:

coleta de dados -> EDA (análise exploratória de dados) -> pré-processamento -> seleção de modelos -> avaliação de desempenho

####Análise exploratória de dados

A análise exploratória de dados, ou Exploratory Data Analysis (EDA) em inglês, é um das etapas do ML onde ocorre, de fato, a análise estatística e o entendimeno dos dados por meio das visualizações geradas a partir dos dados coletados.

Dentre as diversas tarefas comuns, destacam-se: o entendimento do comportamento e distribuição dos dados individualmente ou em conjunto, a identificação de anomalias com inconsistências ou dados faltantes (missing values) e pontos extremos (outliens) e uma possível seleção ou criação de novos atributos ou características (features).

Quanto às visualizações geradas nessa etapa, é importante mencionar que não é requisitado um alto nível de qualidade, pois o foco noEDA é apenas a apresentação e atendimento básico da natureza dos dados e informações que o conjunto de dados possui. Por isso, visando a economia de tempo, podemos apenas gerar gráficos básicos, tomando um caminho mais direto e menos burocrático. Dessa forma, conseguiremos ter uma dimensão rápida dos dados e ainda poderemos responder a perguntas que surgirem no meio do processo de entendimento dos mesmos.

Essencialmente, a etapa da EDA também pode ser subdividido em subetapas, formando um pipeline próprio, o qual vamos nos aprofundar mais.


###Pipeline da EDA

Valemencionar que este diagrama apresentado serve apenas como uma proposta para a prática do EDA. Porém, caso necessário, podem ser utilizadas outras etapas ou até divisões diferentes das que foram adotadas neste material.

load Date -> step1: distinghish -> step2: univariate analysis -> step3: bi-/multivariate analysis -> step4: detect aberrant and missing values -> step5: detect outiliers -> step6: feature engineering


1. Reconhecimento de atributos

Esta etapa consiste no entendimento do dataset demodo geral, relacionando suas dimensões, seus exemplos, representando os registros e entradas, representando suas colunas e características. Os atributos, por sua vez, podem ser investigados quando seu tipo, escala, valores únicos, etc. Para esta etapa, é aconselhável também acessar os dicionários dos dados, caso existam, a fim de ter uma dimensão maior da natureza e detalhes importantes dos dados.

2. Análise univariada

Esta etapa consiste na descrição dos dados avaliando seus atributos, um por vez. Dessa forma, é possível visualizar as distribuições das variáveis e seus respectivos parâmetros estatísticos que os descrevem

3. Análise multivaiada

Esta etapa consiste na descrição dos dados avaliando seus atributos em conjunto. Dessa forma, é possível visualizar correlações entre as variáveis e seus respectivos parâmetros estatísticos que os descrevem.

4. Detecção de anomalias

Esta etapa consiste na identificação de anomalias nos dados, seja referente a inconsistência do valor medido (exemplo: idade acima de 300 anos), seja a audiência de um valor. É importante mencionar mencionar que a ocorrência sistemática de qualquer anomalia pode estar relacionada ao processo de coleta dos dados, algo que, do ponto de vista estatístico, deve ser analisado cuidadosamente e tratada conforme regras específicas, conforme consta no estado da arte.

5. Deleção de outliers

Esta etapa consiste na identificação de pontos extremos que podem representar algum tipo de anomalia nos dados e interferir no parâmetros estatísticos. É importante lembrar que nem sempre um outlier é decorrente de de um erro ou anomalia, pois simplesmente pode representar um espaço pouco explorado ou com poucos exe à disposição no dataset.

6. Engenharia de Features

Esta etapa consiste na transformação e criação de novas características no dataset ou, simplesmente, na diminuição de sua dimensionalidadepor meio da selação das características mais relevantes e as que melhor representam a variância dos dados. Esta etapa está diretamente relacionada com o entendimento da natureza dos dados e do negócio. Porém, vale lembrar que nem sempre esta etapa ocorre dentro da EDA, podendo fazer também parte do pré-processamento.

Observações:

- Uma observação importante nessas etapas é que, em diversos casos, a EDA e o pré-processamento entram num ciclo, onde o que é descoberto em uma etapa é transformado na consequente que, por sua vez, após estas transformações aplicadas, é possível criar novas visualizações e entendimentos complementares sobre os dados originados, novamente na EDA.

- Além disso, as etapas de descoberta de anomalias ou outliners podem ser tratadas ainda na EDA ou, simplesmente, mantidas para que no pré-processamento de dados sejam passadas por uma limpeza e transformação adicinais.
