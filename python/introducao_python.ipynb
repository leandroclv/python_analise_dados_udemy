###Operadores de identidade

Os operadores de identidade são usados para comparar os objetos, não se forem iguais, mas se forem realmente o mesmo objeto, com  a mesma localização de memória:

1. is (retorna True se ambos as variáveis forem o mesmo objeto)
2. is not (retorna True se ambas as variáveis não forem o mesmo objeto)

```python
string_1 = 'Olá Mundão!'
string_2 = 'Olá Mundão'
```

```python
type(string_1) == type(string_2)
```

```python
type(string_1) is not type(string_2)
```

###Operadores de Associação

Os operadores de associação são usados para testar se uma sequeência é apresentada em um objeto:

1. in (retorna True se uma sequência com o valor especificado estiver presente no objeto)
2. not in (retorna True se uma sequência com o valor especificado não estiver presente no objeto) 

```python
lista_acoes = ['Magalu', 'Via', 'Carrefour']
```

```python
'Via' in lista_acoes
```

```python
'zz' in lista_acoes
```

```python
'zz' not in lista_acoes
```

###Manipulando Listas

As listas são usadas para armazenar vários itens em uma única variável.
As listas são um dos 4 tipos de dados internos do Python usados para armazenar coleções de dados, os outros 3 são Tuple, Set e Dicionary, todos com qualidades e usos diferentes.

Comandos mais utilizados:

1. append() - Para adicionar um item ao final da lista
2. len() - Calcular o tamanho da lista
3. [] - Acessar posições
4. del() - Excluir um elemento
5. clear() - Limpa a lista
6. insert() - Para inserir umitem de lista em um índice especificado
7. extend() - Anexar elementos de outra lista à lista atual
8. remove() - Remove o item especificado
9. pop() - Remove o índice especificado
10. sort() - Ordenar os valores 
11. copy() - Faz uma cópia da lista
12. index() - Retorna o index do elemento da lista

```python
lista_vazia = []
```

```python
lista_vazia.append(1)
lista_vazia.append(2)
lista_vazia.append(3)
lista_vazia.append('valor')
print(lista_vazia)
```

```python
len(lista_vazia)
```

```python
#Acessar
lista_vazia[0]
```

```python
lista_vazia[-1]
```

```python
lista_vazia[0:2]
```

```python
print(lista_vazia)
lista_vazia.clear()
print(lista_vazia)
```

```python
lista_vazia = [1,2,3,'valor']
lista_vazia
```

```python
lista_vazia.insert(0, '0') #inserir o '0' na posição 0
print(lista_vazia)
```

```python
lista_vazia.insert(4, 1.3)
print(lista_vazia)
```

```python
lista_1 = [1,2,3]
lista_2 = [4,5,6]
lista_1.extend(lista_2)
lista_1
```

```python
lista_1.remove(5) #valor conhecido
lista_1
```

```python
lista_1.pop(0) #índice
lista_1
```

```python
lista_abc = ['z','c','f','h','a']
lista_abc.sort()
lista_abc
```

```python
lista_abc.sort(reverse=True)
lista_abc
```

```python
lista_nova = lista_abc.copy()
lista_nova
```

```python
lista_nova.index('h')
```

####Datetime

O datetime módulo fornece classes para manipulação de datas e horas

```python
import datetime
```

```python
dia_hoje = datetime.datetime.today()
print(dia_hoje)
```

```python
print(datetime.datetime.today().date())
```

```python
data = datetime.datetime.today().date()
```

```python
ano = data.year
mes = data.month
dia = data.day

print(dia, mes, ano)
```

```python
data_antiga = datetime.date(2026, 6, 16)
data_antiga
```

```python
data - data_antiga
```

```python
data.strftime('%d/%m/%Y')
```

```python
data + datetime.timedelta(days=45)
```
