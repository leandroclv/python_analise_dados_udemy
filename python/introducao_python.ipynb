###Loop while

Python while loop é usado para executar um bloco de instruções repetidamente até que uma determinada condição seja satisfeita. E quando a condição se torna falsa, a linha imediatamente após o loop no programa é executada. while loop se enquadra na categoria de iteração indefinida. A iteração indefinida significa que o número de vezes que o loop é executado não é especificado explicitamente com antecedência.

```python
parar = 0

while parar <= 10:
    print(parar)
    parar += 1
```

```python
loop = 0

while loop <= 10:
    print(loop)
    if loop == 5:
        for x in range(5):
            print(f'{x} for')
            
    loop += 1
```

```python
#jogo
while True:
    tu = random.randint(0,10)
    contra = random.randint(0,10)
    
    print(f'Eu tirei o valor {tu}')
    print(f'E vocês tiraram {contra}')
    print('-' *30)
    
    if tu > contra:
        print('Ganhei')
        break
```

###Break e Continue

O uso de loops no Pyhton automatiza e repete as tarefas de maneira eficiente. Mas, às vezes,pode surgir uma condição em que você deseja sair completamente do loop, pular uma iteração ou ignorar essa condição. Isso pode ser feito por instruções de controle de loop. As instruções de controle de loop alteram a execução de sua sequência normal. Quando a execução sai de um escopa, todos os objetos automáticos que foram criados nesse escopo são destruídos. Python suporta as seguintes instruções de controle.

```python
lista = ['morango', 'pera', 'uva', 'abacaxi']
```

```python
for pais in lista:
    print(pais)
```

```python
for pais in lista:
    print(pais)
    if pais == 'uva':
        break
```

```python
for pais in lista:
    if pais == 'pera':
        continue
    print(pais)
```
