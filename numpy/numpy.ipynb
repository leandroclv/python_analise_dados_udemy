###Numpy

Numpy fornece um grande conjunto de funções e operações de biblioteca que ajudam os programadores a executar facilmente cálculos numéricos.

- 1D array: axis 0 -> shape (4,)
- 2D array: axis 0 e axis 1 -> shape (2,3)
- 3d array: axis 0, axis 1 e axis 2 -> shape (4,3,2)

Mas em Numpy, as dimensões são chamadas de eixos. O número de eixos ẽ classificado. Em termos mais simples, quando você tem mais de uma matriz unidimensiona, então o conceito do eixo surge. Por exemplo, a matriz 2-D tem 2 eixos.
