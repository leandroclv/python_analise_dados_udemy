Sobre o conjunto de dados Este conjunto de dados consiste nas notas obtidas pelos alunos em várias disciplinas.

Introdução Vamos tentar entender a influência dos antecedentes dos pais, preparação para testes etc. no desempenho dos alunos.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('StudentsPerformance(1).csv')
df.head()
```

```python
df.rename(columns = {
    'gender': 'Sexo',
    'race/ethnicity': 'Raça/etinia',
    'parental level of education': 'Nível de educação parental' ,
    '': 'Almoço',
    'test preparation course': 'Curso preparatório',
    'math score': 'Nota de matemática',
    'reading score': 'Nota de leitura',
    'writing score': 'Nota de escrita'
}, inplace=True)
df
```

```python
df['Sexo'].value_counts(normalize=True).round(2) * 100
```
